```python
import yfinance as yf

# Microsoft (MSFT)에 대한 Ticker 객체 생성
msft =yf.Ticker("MSFT")

# 아마존 "AMZN"
# 애플 "AAPL"
# 구글 "GOOGL"
# 삼성 "005930.KQ"

# Ticker 객체의 정보 출력(.py에서 실행할 때는 print(msft.info)로 사용)
display(msft.info)

#Shift + Enter를 눌러서 실행하면, Microsoft에 대한 다양한 정보가 출력됩니다. 이 정보에는 회사 개요, 주식 가격, 재무 데이터, 배당금 정보 등이 포함됩니다.
```

```python
hist = msft.history(period="5d")  # 최근 5일간의 주식 가격 데이터를 가져옵니다.
display(hist)  # 데이터프레임 형태로 출력됩니다.
```

```python
display(msft.recommendations)  # Microsoft에 대한 애널리스트 추천 정보를 출력합니다. 
```
